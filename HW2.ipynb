**1 Cчитайте датасет из файла**

```python
#pip install pandas
#pip install numpy
import numpy as np
import pandas as pd
df = pd.read_excel('Реестр_выполненных_рейсов_за_период.xlsx')
df.columns = df.iloc[0]
df = df.drop(0).reset_index(drop=True)
df = df.drop(11268).reset_index(drop=True)
df = df.drop(11268).reset_index(drop=True)
df = df.drop(11268).reset_index(drop=True)
df.dropna(how='all', axis=1, inplace=True)
df.reset_index().tail()
```

**2 Выведите основную информацию о датасете: информацию о типах данных, число пропусков, средние значения**

---



```python
print ('\033[1m'+"Информация о типе данных"+'\033[0m')
df.info()
print ('\033[1m'+"\n\n Пропущенные значения"+'\033[0m')
print (df.isna().sum())
print ('\033[1m'+"\n\n Средние значения"+'\033[0m')
print (df.describe())
```

**3 Посчитайте процент выполненных рейсов по каждому из аэропортов (Код АП), выведите топ-10 результатов**

---



```python
df.per = df['Код АП'].value_counts(normalize=True).mul(100)
df.per.head(10).reset_index()
```

**4 Посчитайте процент выполненных рейсов ВВЛ по каждому из аэропортов (Код АП), выведите топ-10 результатов**

---



```python
df.perdom = df.loc[df['Направление'] == 'Domestic', 'Код АП'].value_counts(normalize=True).mul(100)
df.perdom.head(10).reset_index()
```

**5 Выведите самый популярный маршрут по каждому из направлений (Domestic/International)**


---



```python
#select = "International" # @param ["Domestic", "International"]
airport_count = df.loc[df['Направление'] == "Domestic", 'Код АП'].value_counts().reset_index()
airport_count.columns = ['Код АП', 'Кол-во рейсов']
airport_max = airport_count['Кол-во рейсов'].max()
airport_best = airport_count[airport_count['Кол-во рейсов'] == airport_max]['Код АП']
print ('\033[1m'+'Самый популярный маршрут Domestic:' +'\033[0m', ','.join(airport_best))

airport_count = df.loc[df['Направление'] == "International", 'Код АП'].value_counts().reset_index()
airport_count.columns = ['Код АП', 'Кол-во рейсов']
airport_max = airport_count['Кол-во рейсов'].max()
airport_best = airport_count[airport_count['Кол-во рейсов'] == airport_max]['Код АП']
print ('\033[1m'+'Самый популярный маршрут International:' +'\033[0m', ','.join(airport_best))

```

**6 Выведите часть таблицы с рейсами категории C и маршрутом САЛЕХАРД**

**7 Выведите на экран результат из пункта выше с сортировкой по убыванию Baggage Weight**

**8 Выведите среднюю МВМ и Вместимость с группировкой по направлению и типу ВС**

---
